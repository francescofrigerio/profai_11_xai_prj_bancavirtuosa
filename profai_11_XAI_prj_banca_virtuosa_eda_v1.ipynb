**HUMAN IN THE LOOP : MIGLIORAMENTO CONTINUO** <br>

Le spiegazioni in termini umani sono un valore economico tangibile perchè
le persone si fidano del modello
come Netflix quando ci dice: “ti consiglio questo film perché hai visto quest'altro”.<br>
Le spiegazioni in termini umani permettono di costruire modelli più robusti
in particolare nella computer vision perchè stiamo spiegando
al modello d'intelligenza artificiale di ragionare in termini umani quando
cerca gli oggetti. <br>

Le spiegazioni in termini umani permettono il cosidetto "Human in the Loop"
cioè l'essere umano diventa un anello di una catena di miglioramento
continuo del modello perchè vediamo l'output del modello e dove sbaglia
e possiamo correggere un hiperparametro e ridiamo il tutto di nuovo in pasto al modello. <br>
Se il modello oltre a dare le previsioni ci dice il perchè delle sue previsioni questo rappresenta un loop con feed back
continuo tra umano e modello è molto + efficace e ci da dei modelli veramente + performanti. <br>
