4.**SUPER PIXEL** <br>
Nel contesto dell'Explainability (XAI) per modelli di Machine Learning e Deep Learning applicati alle immagini,
un superpixel è un gruppo coerente di pixel adiacenti che condividono
caratteristiche visive simili (come colore, intensità o texture).

Anziché analizzare l'immagine pixel per pixel — un processo computazionalmente pesante e poco interpretevole
per gli esseri umani — il modello XAI suddivide l'immagine in queste regioni
"dotate di senso". <br>


I metodi di spiegazione (LIME,SHAP,GRAD CAM ecc.)
 usano i superpixel per spiegare le decisioni delle reti neurali visive: <br>

 Riduzione della complessità: Un'immagine da $224 x 224$ contiene
 oltre 50.000 pixel, ma può essere riassunta in circa 50–100
 superpixel significativi. <br>

 Interpretabilità umana: Gli esseri umani non riconoscono un oggetto da un singolo pixel
 isolato, ma da forme e regioni (es. il muso di un cane, una ruota di un'auto). <br>

 Perturbazione delle regioni: Per capire quali parti dell'immagine influenzano la previsione,
 alcuni algoritmi di spiegazione (Occlusion e Lime) "nascondono" o disattiva (maschera)
 intere regioni/superpixel e osserva come cambia l'output della rete. <br>

 Esempio pratico con LIME <br>
 Se una rete neurale classifica un'immagine come "Cane":
 L'algoritmo di segmentazione divide l'immagine in superpixel (es. muso, orecchie, sfondo erboso). <br>
 LIME genera varianti dell'immagine coprendo casualmente alcuni superpixel.
 Calcola il punteggio di importanza per ciascun superpixel. <br>
 Risultato visivo: Evidenzia i superpixel che hanno contribuito positivamente alla decisione
  (es. la regione con il muso del cane) e quelli neutri o negativi (es. l'erba dello sfondo). <br>

6. **SHAP** <br>

SHAP è locale e globale allo stesso tempo. <br>

La spiegazione SHAP di un singolo campione è locale.
DenseNet riceve un'immagine MNIST e predice 8.
SHAP attribuisce un valore a ciascuna feature, indicando quanto questa contribuisce
alla predizione rispetto a un valore di riferimento. <br>

I valori SHAP possono essere positivi o negativi. <br>
Positivo: contribuisce ad aumentare l'output considerato. <br>
Negativo: contribuisce a diminuirlo.  <br>

La spiegazione locale riguarda una specifica osservazione. <br>
La spiegazione globale si ottiene aggregando le spiegazioni locali su un insieme di osservazioni. <br>

La coalizione NON è il singolo sample <br>

bisogna distinguere tra  <br>
-Il singolo sample da spiegare. <br>
-Le coalizioni di feature. <br>
-Il background dataset. <br>

Supponiamo di avere un'immagine MNIST 28×28 quindi 784 pixel. <br>

SHAP vuole capire il contributo di ciascun pixel alla predizione. <br>
Considera differenti combinazioni di feature presenti o assenti e valuta il comportamento del modello. <br>

-Il sample da spiegare è l'osservazione sulla quale stai calcolando l'attribuzione. <br>

-Una coalizione è quindi la differente combinazione di feature presenti o assenti quindi
 un sottoinsieme delle feature. <br>

-il background dataset serve a definire il riferimento rispetto al quale valutare il contributo delle feature. <br>

Per immagini, spesso si preferisce raggruppare i pixel in superpixel o regioni,
perché utilizzare direttamente 784 feature può essere computazionalmente
oneroso e rendere difficile l'interpretazione. <br>
